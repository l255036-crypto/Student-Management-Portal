### Import Libraries & Define CSV Fields

```python
import csv

file_name= "students_new.csv"
```

Explanation:
Imported Python's built-in csv module and stored the data file name (students_new.csv) in a variable called file_name. The csv module lets us save and read student data in a table-like file. Keeping the file name in one variable means we only change it in one place if needed.

```python
fields = [
    "id", "name", "age",
    "department", "degree",
    "section", "email",
    "course1", "course2",
    "course3", "course4",
    "course5", "gpa"
]
```

Explanation: We made a list called fields with the 13 column names: id, name, age, department, degree, section, email, five course columns and gpa.
The list works as the header row of the CSV file. It also keeps the column order the same every time we write to the file.

### Load Records Function

```python
def load_records():
    students = []

    try:
        with open(file_name, "r", newline="") as file:

            reader = csv.DictReader(file)

            for row in reader:
                students.append(row)

        print("Student records loaded successfully!")

    except FileNotFoundError:
        print("Database not found. Starting with empty records.")

    except PermissionError:
        print("Permission denied while reading the file.")

    except csv.Error as error:
        print("Error reading CSV file:", error)

    return students
```

Explanation:
This function opens the CSV file, reads every row with csv.DictReader and adds each row to a list called students. It returns that list.The assignment says the portal must read old records when it starts. We used a for loop to go through the rows and try-except to handle FileNotFoundError (file missing), PermissionError (no access) and csv.Error (bad file content), so the program never crashes.

### Save Records Function

```python
def save_records(students):

    try:
        with open(file_name, "w", newline="") as file:

            writer = csv.DictWriter(
                file,
                fieldnames=fields
            )

            writer.writeheader()
            writer.writerows(students)

        print("Records saved successfully.")
        return True

    except PermissionError:
        print("Permission denied while saving records.")
        return False
```

Explanation: This function writes the header and all student records back into the CSV file using csv.DictWriter. It returns True if saving worked and False if it failed. Every add, update and delete must be stored permanently. Returning True or False lets the other functions know if the save worked, so they can undo a change if it did not. PermissionError is handled with try-except.
We loaded the records, printed them and saved them back.
We did this to test that both file functions work together before building the rest of the portal.
Output: "Student records loaded successfully!" and an empty list [], because no student was added yet. Then "Records saved successfully." and True, which is the return value of save_records().

### Add Student Function

```python
def add_record(students):
    print("ADD NEW STUDENT")

    student_id = input("Enter Student ID: ")
    for student in students:
        if student["id"] == student_id:
            print("Student ID already exists.")
            return students

    name = input("Enter Student Name: ")
    department = input("Enter Department: ")
    degree = input("Enter Degree: ")
    section = input("Enter Section: ")
    email = input("Enter Email: ")

    if not all([student_id, name, department,
                degree, section, email]):
        print("All personal information fields are required.")
        return students

    try:
        age = int(input("Enter Age: "))
        if age <= 0:
            print("Age must be greater than zero.")
            return students
    except ValueError:
        print("Please enter a valid age.")
        return students

    try:
        number = int(input("How many courses? (1-5): "))
        if number < 1 or number > 5:
            print("You can enter between 1 and 5 courses.")
            return students
    except ValueError:
        print("Please enter a valid number.")
        return students
    courses = []

    for i in range(number):
        course_name = input("Enter Course " + str(i + 1) + ": ")
        if not course_name:
            print("Course name cannot be empty.")
            return students
        courses.append(course_name)

    try:
        gpa = float(input("Enter Final GPA (0-4.00): "))
        if gpa < 0 or gpa > 4:
            print("GPA must be between 0.00 and 4.00.")
            return students

    except ValueError:
        print("Please enter a valid GPA.")
        return students

    new_student = {
        "id": student_id,
        "name": name,
        "age": age,
        "department": department,
        "degree": degree,
        "section": section,
        "email": email,
        "gpa": round(gpa, 2)
    }
    for i in range(5):
        if i < len(courses):
            new_student["course" + str(i + 1)] = courses[i]
        else:
            new_student["course" + str(i + 1)] = ""

    students.append(new_student)

    if save_records(students):
        print("\nStudent added successfully!")
    else:
        students.pop()
        print("Student could not be saved.")

    return students
```

Explanation: The function takes the student list, asks for all details of a new student, checks them, adds the student to the list and saves the file. It returns the updated list. This is the main "create" feature. It checks that the ID is not a duplicate, no field is empty, age is a positive whole number, the number of courses is between 1 and 5 and the GPA is between 0 and 4. A for loop collects the course names, and try-except catches ValueError when a number is typed wrongly. If saving fails, the new student is removed again with pop().

### View Students Function

```python
def view_records(students):

    print("STUDENT RECORDS")

    if len(students) == 0:
        print("No student records available.")
        return

    for student in students:

        print("\nStudent ID:", student["id"])
        print("Name:", student["name"])
        print("Age:", student["age"])
        print("Department:", student["department"])
        print("Degree:", student["degree"])
        print("Section:", student["section"])
        print("Email:", student["email"])
        print("Registered Courses:")

        for i in range(1, 6):
            course_name = student["course" + str(i)]
            if course_name != "":
                print(str(i) + ".", course_name)

        print("Final GPA:", student["gpa"])
```

Explanation: The function prints every student's details one by one. If the list is empty, it prints a message instead. This step is done because the admin needs a way to see all stored records. We used a for loop for the students and another for loop for the five course columns, and it only prints courses that are not empty.

### Search Student Function

```python
def search_record(students):

    print("SEARCH STUDENT")

    student_id = input("Enter Student ID: ")

    for student in students:
        if student["id"] == student_id:
            print("Student Found")

            print("\nStudent ID:", student["id"])
            print("Name:", student["name"])
            print("Age:", student["age"])
            print("Department:", student["department"])
            print("Degree:", student["degree"])
            print("Section:", student["section"])
            print("Email:", student["email"])

            print("\nRegistered Courses:")
            for i in range(1, 6):
                course = student["course" + str(i)]
                if course != "":
                    print(str(i) + ".", course)

            print("\nFinal GPA:", student["gpa"])
            return student

    print("Student not found.")
    return None
```

Explanation: The function asks for a Student ID and loops through the list to find a match. If found, it prints the full record and returns it. If not, it prints "Student not found." and returns None.
Searching by ID is the fastest way to find one student, because every ID is unique.

### Update Student Function

```python
def update_record(students):

    print("UPDATE STUDENT RECORD")
    student_id = input("Enter Student ID: ")

    for student in students:
        if student["id"] == student_id:
            print("Student found:", student["name"])
            print("\nWhat would you like to update?")
            print("1. Name")
            print("2. Age")
            print("3. Department")
            print("4. Degree")
            print("5. Section")
            print("6. Email")
            print("7. Courses")
            print("8. GPA")

            choice = input("Enter your choice: ")

            if choice == "1":
                student["name"] = input("Enter new name: ")
            elif choice == "2":
                try:
                    age = int(input("Enter new age: "))
                    if age <= 0:
                        print("Age must be greater than zero.")
                        return students
                    student["age"] = age
                except ValueError:
                    print("Please enter a valid age.")
                    return students

            elif choice == "3":
                student["department"] = input("Enter new department: ")
            elif choice == "4":
                student["degree"] = input("Enter new degree: ")
            elif choice == "5":
                student["section"] = input("Enter new section: ")
            elif choice == "6":
                student["email"] = input("Enter new email: ")
            elif choice == "7":

                try:
                    number = int(input("How many courses? (1-5): "))
                    if number < 1 or number > 5:
                        print("Enter a number between 1 and 5.")
                        return students
                except ValueError:
                    print("Please enter a valid number.")
                    return students

                for i in range(1, 6):
                    student["course" + str(i)] = ""

                for i in range(1, number + 1):
                    course = input("Enter Course " + str(i) + ": ")
                    student["course" + str(i)] = course

            elif choice == "8":

                try:
                    gpa = float(input("Enter new GPA (0-4.00): "))
                    if gpa < 0 or gpa > 4:
                        print("GPA must be between 0.00 and 4.00.")
                        return students
                    student["gpa"] = round(gpa, 2)
                except ValueError:
                    print("Please enter a valid GPA.")
                    return students

            else:
                print("Invalid choice.")
                return students

            if save_records(students):
                print("\nStudent record updated successfully!")
            else:
                print("Could not save the changes.")
            return students

    print("Student not found.")
    return students
```

Explanation: The function finds a student by ID, shows a menu of fields (name, age, department, degree, section, email, courses, GPA), takes the new value and saves the file.
This is necessary because student details change over time, so the admin needs to edit them without deleting and re-adding the student. Age, course count and GPA are checked again with try-except so wrong values are rejected. We called add_record() with a test student (ID 1) to check that the function works. Output: The prompts ask for each detail, then "Records saved successfully." and "Student added successfully!" appear. This means the student was added and written to the CSV file.

### Delete Student Function

```python
def delete_record(students):

    print("DELETE STUDENT RECORD")
    student_id = input("Enter Student ID to delete: ")

    for i in range(len(students)):
        if students[i]["id"] == student_id:
            print("Student found")
            print("Name:", students[i]["name"])
            print("Student ID:", students[i]["id"])

            confirm = input("Are you sure you want to delete this student? (yes/no): ")

            if confirm == "yes":
                deleted_student = students.pop(i)
                if save_records(students):
                    print("Student record deleted successfully.")
                else:
                    students.insert(i, deleted_student)
                    print("Deletion unsuccessful.")
            else:
                print("Deletion cancelled.")
            return students

    print("Student not found.")
    return students
```

Explanation: The function finds a student by ID, shows the name and ID, and asks for confirmation. Only the answer "yes" deletes the record. Then it saves the file.
Deleting is permanent, so we added a confirmation to avoid mistakes. If saving fails, the student is put back with insert(), so the list and the file stay in line.

### Generate Academic Report Function

```python
def generate_report(students):

    print("ACADEMIC REPORT:")

    if len(students) == 0:
        print("No student records available.")
        return None

    total_gpa = 0
    highest_gpa = 0
    lowest_gpa = 4
    highest_student = ""
    lowest_student = ""
    count = 0

    for student in students:
        try:
            gpa = float(student["gpa"])
            total_gpa += gpa
            count += 1
            if gpa > highest_gpa:
                highest_gpa = gpa
                highest_student = student["name"]
            if gpa < lowest_gpa:
                lowest_gpa = gpa
                lowest_student = student["name"]
        except (ValueError, TypeError):
            print("Invalid GPA for:", student["name"])

    if count == 0:
        print("No valid GPA records available.")
        return None

    average_gpa = total_gpa / count

    print("\nTotal Students:", len(students))
    print("Students with valid GPA:", count)
    print("Average GPA:", round(average_gpa, 2))
    print("\nHighest GPA:", highest_gpa)
    print("Student:", highest_student)
    print("\nLowest GPA:", lowest_gpa)
    print("Student:", lowest_student)

    report = {
        "total_students": len(students),
        "average_gpa": round(average_gpa, 2),
        "highest_gpa": highest_gpa,
        "highest_student": highest_student,
        "lowest_gpa": lowest_gpa,
        "lowest_student": lowest_student
    }
    return report
```

Explanation: The function loops through all students, adds up their GPAs, and finds the average, the highest GPA and the lowest GPA with the student names. It returns the results as a dictionary.
This step is done so that the admin gets a quick summary of academic performance. try-except with ValueError skips any record whose GPA cannot be converted to a number. If there are no records, it prints a message and returns None.

### Main Menu Function

```python
def main():

    print("WELCOME TO STUDENT MANAGEMENT PORTAL")
    login_status = user_login()
    if login_status == False:
        print("Access denied.")
        return

    students = load_records()

    while True:
        print("\nMAIN MENU")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Generate Academic Report")
        print("7. Exit")

        choice = input("\nEnter your choice: ")
        if choice == "1":
            students = add_record(students)
        elif choice == "2":
            view_records(students)
        elif choice == "3":
            search_record(students)
        elif choice == "4":
            students = update_record(students)
        elif choice == "5":
            students = delete_record(students)
        elif choice == "6":
            generate_report(students)
        elif choice == "7":
            print("Thank you for using our Student Portal")
            break
        else:
            print("Invalid choice. Please try again.")
```

Explanation: main() joins everything together. It shows a welcome message, runs the login, loads the records and then starts a while True loop with a menu of 7 options. Each option calls one of our functions. Option 7 exits the loop.
The assignment needs a console menu that keeps running until the user chooses to exit. This is also the one function that makes the whole portal work as a single program.
